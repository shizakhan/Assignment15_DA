```python
import pandas as pd
import sqlite3 as sql
```

```python
conn = sql.connect('factbook.db')
conn
```

```python
def sql_query(q):
    return pd.read_sql_query(q,conn)
```

```python
q = 'Select * from facts'
sql_query(q)
```

```python
q = 'Select * from cities'
sql_query(q)
```

### 1. List all cities where the area is greater than 5000, along with their population.

```python
q = ''' select c.name city_name,f.area,f.population from cities c inner join facts f
        on
        c.facts_id = f.id
        where f.area > 5000
'''
sql_query(q)
```

### 2. List the names of capital cities where the death rate is greater than 2% and the birth rate is less than 20%

```python
q = '''select c.name city_name,f.death_rate,f.birth_rate from cities c inner join facts f
        on
        c.facts_id = f.id
        where f.death_rate > 2
        and
        f.birth_rate < 20
'''
sql_query(q)
```

### 3. List the cities with a population growth rate between 1% and 5%, sorted in desc by population growth.

```python
q = '''select c.name city_name, f.population_growth from cities c 
        inner join facts f on
        c.facts_id = f.id
        where f.population_growth between 1 and 5
        order by f.population_growth desc
'''
sql_query(q)
```

### 4. Find countries that start with &quot;A&quot; and do not have any cities listed in the cities table. The query should return the country name and city name. Using left join

```python
q = '''select f.name country_name, c.name city_name from facts f left join 
        cities c on 
        f.id = c.facts_id
        where f.name like "A%"
        and
        c.name is null
'''
sql_query(q)
```

### 5. List all capital cities and their corresponding death rate using a LEFT JOIN.

```python
q = '''select c.name city_name, c.capital, f.death_rate from cities c left join 
        facts f on 
        c.facts_id = f.id
        where c.capital == 1
'''
sql_query(q)
```
